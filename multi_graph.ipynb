```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

fig, axs = plt.subplots(6, 1, figsize=(10, 25))

data1 = np.arange(1, 10)
data2 = np.sin(np.random.randint(1, 10, 9))
data3 = np.random.rand(9)
data4 = np.random.randint(1, 10, 9)

# 1. Line Graph
axs[0].plot(data1, data2, color='blue', marker='*',label='Sine Wave')
axs[0].set_title('Line Graph')
axs[0].set_xlabel('X-axis')
axs[0].set_ylabel('Y-axis')
axs[0].legend()

# 2. Bar Chart
axs[1].bar(data1, data4, color='pink')

axs[1].set_title('Bar Chart')
axs[1].set_xlabel('X-axis')
axs[1].set_ylabel('Values')
axs[1].legend()

# 3. Stacked Bar Chart
axs[2].bar(data1, data4, color='red', label='Data 4')
axs[2].bar(data1, data3, bottom=data4, color='orange', label='Data 3')
axs[2].set_title('Stacked Bar Chart')
axs[2].set_xlabel('X-axis')
axs[2].set_ylabel('Values')
axs[2].legend()



# 4. Histogram
axs[3].hist(data4, bins=5, color='purple', edgecolor='black')
axs[3].set_title('Histogram')
axs[3].set_xlabel('Bins')
axs[3].set_ylabel('Frequency')

# 5. Scatter Plot
axs[4].scatter(data1, data4, color='yellow', marker='x')
axs[4].set_title('Scatter Plot')
axs[4].set_xlabel('X-axis')
axs[4].set_ylabel('Random Values')

axs[5].axis('off')

plt.tight_layout()
plt.show()
```
